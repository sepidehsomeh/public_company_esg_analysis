# Section 3

## Data Cleaning and Transformation


### Remove Non-Analytical Columns

The `logo` and `weburl` columns contain company URLs and are not required for the ESG analysis. Therefore, they are removed from the analytical dataset. The original raw dataset remains unchanged.


```python
df = df.drop(columns=["logo", "weburl"])

df.shape
```

### Handle Missing Industry Values

The `industry` column contains 13 missing values. Since industry is important for ESG analysis, these records are retained and the missing values are labelled as `Unknown`.

```python
df["industry"] = df["industry"].fillna("Unknown")

df["industry"].isnull().sum()
```

### Convert Date Column

The `last_processing_date` column is converted from text to datetime format to ensure the date is stored in an appropriate data type.


```python
df["last_processing_date"] = pd.to_datetime(
    df["last_processing_date"],
    format="%d-%m-%Y",
    errors="coerce"
)

df["last_processing_date"].dtype
```

```python
df["last_processing_date"].isnull().sum()
```

```python
for column in [
    "environment_level",
    "social_level",
    "governance_level",
    "total_level"
]:
    print(column, df[column].unique())
```

```python
for column in [
    "environment_grade",
    "social_grade",
    "governance_grade",
    "total_grade"
]:
    print(column, df[column].unique())
```

## Export Cleaned ESG Data

The cleaned dataset is exported as a new CSV file for further analysis and use in Power BI. The original raw dataset remains unchanged.
